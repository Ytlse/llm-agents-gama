```python
import pandas as pd

# Chargement du fichier JSONL
df = pd.read_json('../../experiments/current/llm_errors.jsonl', lines=True)
#df = pd.read_json('../../experiments/2026-05-13_11_28/llm_errors.jsonl', lines=True)

# Conversion de la colonne time en datetime pour d'éventuelles analyses temporelles
df['time'] = pd.to_datetime(df['time'])
df = df.sort_values(by='provider')
print("Nb duplicates = ", df.duplicated().count())

# Création d'une table de synthèse : Compte des types d'erreurs par provider

summary = pd.DataFrame(df.groupby(['provider', 'error_type','error_message' ]).size().sort_values(ascending=False).reset_index(name='count'))

summary = summary[summary['count'] > 1]

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
display(summary.head(20))

```
