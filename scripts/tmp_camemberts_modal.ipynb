# Répartition modale — 3 camemberts

Comparaison des parts modales : agents LLM (prompt initial), agents LLM (prompt calibré), et enquête mobilité Cerema.

Code couleur officiel du projet : voiture = red, marche = cyan, transports collectifs = green, vélo = purple.

```python
import matplotlib.pyplot as plt

# Modes et palette officielle du projet
modes = ["Voiture", "Marche", "Transports collectifs", "Vélo"]
colors = ["red", "cyan", "green", "purple"]

# Donnees (%)
data = {
    "Choix LLM initial": [57.2, 5.1, 11.0, 26.8],
    "Choix LLM prompt calibre": [63.9, 10.9, 16.8, 8.3],
    "Enquete mobilite Cerema": [56.7, 26.8, 12.4, 4.1],
}

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for ax, (title, values) in zip(axes, data.items()):
    ax.pie(
        values,
        labels=modes,
        colors=colors,
        autopct="%1.1f%%",
        startangle=90,
        counterclock=False,
        wedgeprops={"edgecolor": "white", "linewidth": 1},
    )
    ax.set_title(title, fontsize=13, fontweight="bold")
    ax.axis("equal")

fig.suptitle("Repartition modale des deplacements", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```
